```python
import geopandas as gpd

from ocab.config import Config
```

## CAMELS

```python
cfg_camels = Config('../CAMELS/config_CAMELS_v200.yml')

root_old = 9400
root_new = 10000
```

### Stations

```python
json_file = cfg_camels.path_gis / 'stations.geojson'
col = 'gauge_id'

# load json file
stations = gpd.read_file(json_file)
stations.rename(columns={'id': col}, inplace=True)
stations.sort_values(by=col, inplace=True)

# reindex stations in Catalunya
stations[col] = [
    (ID - root_old + root_new) if root_old <= ID < (root_old + 100) 
    else ID 
    for ID in stations[col]
]

# export reindexed file
stations.to_file(json_file)
```

### Basins

```python
shp_file = cfg.path_gis / 'camelses_basins.shp'
col = 'gauge_id'

# load shapefile
basins = gpd.read_file(shp_file)
basins.sort_values(by=col, inplace=True)

# reindex the basins in Catalunya
basins[col] = [
    (ID - root_old + root_new) if root_old <= ID < (root_old + 100) 
    else ID 
    for ID in basins[col]
]

# export reindexed file
basins.to_file(shp_file)
```
