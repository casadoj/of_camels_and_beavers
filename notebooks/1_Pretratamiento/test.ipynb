```python
path_camels = Path('/mnt/c/Datasets/CAMELS-ES')
```

```python
version= 'v110'
camels = gpd.read_file(path_camels / version / 'shapefiles/camelses/camelses_basin_shapes.shp').set_index('gauge_id')
camels.index = [int(x.split('_')[1]) for x in camels.index]
camels.index.name = 'COD_HIDRO'
camels.head()
```

```python
anuario = pd.read_csv(
    path_camels / 'raw' / '2020-2021' / 'estaciones' / 'Situac 4_Rio.csv',
    sep=';',
    index_col='COD_HIDRO', 
    encoding='latin-1'
    )

mask_regimen = anuario.REGIMEN_RIO.str.lower() == 'natural'
mask_serie = (anuario.ANO_FIN_MEDIDAS.isnull()) |  (anuario.ANO_FIN_MEDIDAS > 1995)
anuario = anuario[mask_regimen & mask_serie]

print(len(anuario))
anuario.head()
```
